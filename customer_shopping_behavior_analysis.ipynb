```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/customer_shopping_behavior.csv")
```

```python
df_raw = df.copy()
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_percentage.sort_values(ascending=False)
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df.shape
```

```python
numeric_columns = [
    "Age",
    "Purchase Amount (USD)",
    "Review Rating",
    "Previous Purchases"
]
```

```python
for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
categorical_columns = df.select_dtypes(include="str").columns

for col in categorical_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
for col in df.select_dtypes(include="str").columns:
    df[col] = df[col].str.strip()
```

```python
df["Gender"] = df["Gender"].str.strip().str.title()
```

```python
df["Gender"].value_counts()
```

```python
yes_no_columns = [
    "Subscription Status",
    "Discount Applied",
    "Promo Code Used"
]

for col in yes_no_columns:
    df[col] = df[col].str.strip().str.title()
```

```python
df["Subscription Status"].value_counts()
```

```python
df["Frequency of Purchases"] = df["Frequency of Purchases"].replace({
    "Bi weekly": "Bi-Weekly",
    "Fortnightly": "Bi-Weekly"
})
```

```python
df["Frequency of Purchases"].value_counts()
```

```python
df["Payment Method"] = df["Payment Method"].replace({
    "Pay Pal": "PayPal",
    "CreditCard": "Credit Card"
})
```

```python
df["Payment Method"].value_counts()
```

```python
df["Shipping Type"] = df["Shipping Type"].replace({
    "Free shiping": "Free Shipping",
    "Store pick-up": "Store Pickup"
})
```

```python
df["Review Rating"] = df["Review Rating"].fillna(
    df["Review Rating"].median()
)
```
